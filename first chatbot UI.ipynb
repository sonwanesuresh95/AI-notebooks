# chatbot with gradio UI

```python
import os
import gradio as gr
from openai import OpenAI
from dotenv import load_dotenv
load_dotenv()
```

```python
client = OpenAI()
def llm_chat(message, history):
    messages_payload = history + [{"role": "user", "content": message}]
    stream = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages_payload,
        stream=True
    )
    
    response_text = ""
    for chunk in stream:
        if chunk.choices[0].delta.content is not None:
            response_text += chunk.choices[0].delta.content
            yield response_text
```

```python
demo = gr.ChatInterface(
    fn=llm_chat,
    title="My Chatbot",
    description="Ask anything to gpt-4o-mini with streaming responses"
)
demo.launch()
```

gradio takes care of messages(str) and history(dict) internally

```python
gr.close_all()
```

# chatbots with context - a clothes store chatbot assistant

```python
client = OpenAI()
def llm_chat(message, history):
    system_prompt = """
    You are helpful assistant in a clothes store. You should try gently encouraging customers to try items that are on sale.
    Hats are 60% off and most of the other items are 50% off. Encourage cutomers to buy hats if they are unsure to what to buy.
    """
    system_message = [{
        "role": "system",
        "content": system_prompt
    }]
    messages_payload = system_message + history + [{"role": "user", "content": message}]
    stream = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages_payload,
        stream=True
    )
    
    response_text = ""
    for chunk in stream:
        if chunk.choices[0].delta.content is not None:
            response_text += chunk.choices[0].delta.content
            yield response_text
```

```python
demo = gr.ChatInterface(
    fn=llm_chat,
    title="My Chatbot",
    description="Ask anything to gpt-4o-mini with streaming responses"
)
demo.launch()
```

```python
gr.close_all()
```
