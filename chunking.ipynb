```python
import wikipedia
```

# load document

```python
def load_document_from_wikipedia(title: str) -> str:
    """Fetches a Wikipedia page and returns its title and content as a single string."""
    try:
        page = wikipedia.page(title, auto_suggest=False)
        return f"{page.title}\n{page.content}"
        
    except (DisambiguationError, PageError) as e:
        raise ValueError(f"Could not retrieve page '{title}' from Wikipedia API.") from e
```

```python
page = load_document_from_wikipedia("Porsche 911")
print(page[:500])
```

# chunking
- recursive character text splitter chunking - Industry standard method for chunking
  - (default) general texts/articles/documents: chunk size: 500 to 1000 | overlap: 50 to 100 (10%-20% of chunk size)
  - hyper-specific retrieval - legal/medical data: chunk size: 200 to 500 | overlap: 20 to 50 (10%-20% of chunk size)

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
def chunk_document(
    document: str, 
    chunk_size: int = 1000, 
    chunk_overlap: int = 100) -> List[str]:
    """Splits a raw string document into smaller string chunks."""
    if not document.strip():
        return []

    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        length_function=len,
        separators=["\n\n", "\n", " ", ""]
    )
    return text_splitter.split_text(document)
```

```python
chunks = chunk_document(page, chunk_size=1000, chunk_overlap=100)
```

```python
for chunk in chunks[:5]:
    print(chunk)
    print("------------------------------------------------------")
```
